```python
def blur_image(i:int,j:int,array:npt.NDArray[np.floating])->npt.NDArray[np.floating]:
    if i>0 and j>0:
        i-=1
        j-=1
        top = array[0+i:i+2,1+j:3+j]
        left = array[i+1:i+3,j+0:j+2]
        bottom = array[i+2:4+j,j+1:j+3]
        right = array[i+1:i+3,2+j:j+4]
        avg = (top+left+bottom+right)/5
        return avg
```

```python
blur_image(1,1,m441)
```

```python
def smooth_image(img:npt.NDArray[np.floating])->npt.NDArray[np.floating]:
    top = img[:-2,1:-1]
    left = img[1:-1,:-2]
    bottom = img[2:,1:-1]
    right = img[1:-1,2:]
    center = img[1:-1,1:-1]
    blured_img = (top+left+bottom+right+center)/5
    return blured_img
```

```python
smoothed_image:npt.NDArray[np.floating] = image.copy()
for i in range(100): smoothed_image = smooth_image(smoothed_image)
plt.figure()
plt.subplot(1,3,1)
plt.title('Original')
plt.imshow(image,cmap=plt.cm.hot)

plt.subplot(1,3,2)
plt.title('Smoothed')
plt.imshow(smoothed_image,cmap=plt.cm.hot)

plt.subplot(1,3,3)
plt.title("Difference")
plt.imshow(image[1:-1,1:-1]-smooth_image(image))
plt.show()
```
